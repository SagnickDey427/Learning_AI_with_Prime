```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://quotes.toscrape.com/page/"
quotesList=[]
authorsList=[]
for i in range(1,11):
    try:
        resp = requests.get(f"https://quotes.toscrape.com/page/{i}/")
        resp.raise_for_status()
        if resp.status_code == 200:
            htmlCode = resp.text
            soup = BeautifulSoup(htmlCode, "lxml")
            allQuotes = soup.find_all("div",class_="quote")
            for quote in allQuotes:
                tagElements = quote.find('div',class_='tags').find_all('a',class_='tag')
                tags=[]
                for tagElement in tagElements:
                    tags.append(tagElement.text)
                if 'life' in tags:
                    quotesList.append(quote.find('span',class_='text').text)
                    author = quote.find("small",class_="author")
                    authorsList.append(author.text)
            # print(authorsList)
            # print(quotesList)
            
    except requests.exceptions.RequestException as e:
        print(e)

quotesDict ={
    "quotes":quotesList,
    "authors":authorsList
}

df = pd.DataFrame(quotesDict)
df.to_csv('life_quotes.csv')
df
    
```
