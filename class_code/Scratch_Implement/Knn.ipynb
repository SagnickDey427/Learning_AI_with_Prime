```python
import numpy as np
import seaborn as sns

class KNNClassifier:
    def __init__(self,k=3):
        self.k = k
    def __euclid_dist(self,x,y):
        return np.sqrt(np.sum((x-y)**2))
    def fit(self,X,y):
        self.X_train = X
        self.y_train = y
    def __predict_one(self,x):
        knn_dist = [self.__euclid_dist(x,item) for item in self.X_train]
        knn_nbs = np.argsort(knn_dist)[:self.k]
        knn_classes = [self.y_train[i] for i in knn_nbs]
        maj_class = np.argmax(np.bincount(knn_classes))
        return maj_class
    def predict(self,X):
        y_pred = [self.__predict_one(x) for x in X]
        return np.array(y_pred)
```

```python
X_train = np.array([
    [1,2],
    [2,3],
    [3,3],
    [6,5],
    [7,7]
])

y_train = np.array([0,0,0,1,1])

X_test = np.array([
    [2,2],
    [6,6]
])


knnModel = KNNClassifier()
knnModel.fit(X_train,y_train)
y_pred = knnModel.predict(X_test)

print(y_pred)
```
