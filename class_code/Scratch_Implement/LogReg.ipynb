```python
import numpy as np

class LogisticRegression:
    def __init__(self, learning_rate=0.01, n_itr=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_itr

    def __sigmoid(self,z):
        return 1/(1+ np.exp(-z))
    
    def fit(self, X, y): #X_train, y_train
        m,n = X.shape

        #Step 1 : initialise params
        self.bias = 0
        self.weights = np.zeros(n)

        for i in range(self.n_iter):
            #Step 2 : find predicted val
            z = self.bias + np.dot(X,self.weights)
            y_pred = self.__sigmoid(z)
    
            #Step 3 : Calculate slope
            db = (1/m) *np.sum(y_pred-y)
            dw = (1/m) * np.dot(X.T,(y_pred-y))
    
            #Step 4 : Convergence theorem -- update params
            self.bias = self.bias - self.lr*db
            self.weights = self.weights - self.lr*dw
    def probabilities(self,X):
        z = self.bias + np.dot(X,self.weights)
        return self.__sigmoid(z)

    def predict(self,X,threshold=0.5):
        z =  self.probabilities(X)
        return (z>=threshold).astype(int)
        
```

```python
X = np.array([[1,2],[2,3],[3,4],[4,5]])
y = np.array([0,0,1,1])

model = LogisticRegression()
model.fit(X,y)

y_pred = model.predict(X)
print(y_pred)
print(model.bias)
print(model.probabilities(X))
```
