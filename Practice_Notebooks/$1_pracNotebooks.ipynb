```python
df = pd.read_csv("../DataSets/raw_data.csv")
df.isna().sum()
clean_df = df.copy()
df
# df.dtypes
```

```python
clean_df['age'] =  clean_df['age'].fillna(clean_df['age'].mean())
clean_df['country'] = clean_df['country'].ffill()
clean_df['name'] = clean_df['name'].bfill()
clean_df['gender'] = clean_df['gender'].fillna(df['gender'].mode()[0])
clean_df['income'] =  clean_df['income'].fillna(clean_df['income'].median())
clean_df
# clean_df.isna().sum()
```

```python
clean_df.duplicated().sum()
clean_df2 = clean_df.drop_duplicates()
# clean_df2['age'].astype("int64")
clean_df3 = clean_df2.copy()
clean_df3['age'] = clean_df3['age'].astype("int64")
clean_df3
```

```python
df2 = clean_df3.copy()
df2['tax'] = df2['income'].apply(lambda x : "25%" if x >= 50000 else "15%")
df2
```

```python
gender_map = {"Male":"M", "Female":"F","Unknown":"U"}
df2['gender'] = df2['gender'].map(gender_map)
df2
```

```python
df2 = df2.assign(income =lambda x : x['income'] * 1.1)
df2
```

```python
df2['country'] = df2['country'].replace('USA','US')
df2
```

```python
copy_df = df2.copy()
columns = ["id","name","age","country","gender","income","tax"]
nw_cols = list(filter(lambda x : x != "id" , columns)) 
nw_cols.append("id")
# nw_cols

copy_df[nw_cols]
```
