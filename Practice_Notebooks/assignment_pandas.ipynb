```python
df = pd.read_csv("../DataSets/IRIS.csv")
```

```python
# df.sample(5)
# df.head(10)
# df.shape
# df.dtypes
# df.describe()
# df.query("petal_length > 4.5 and species == 'Iris-virginica'")
# df[(df["petal_length"] > 4.5) & (df["species"] == 'Iris-virginica')]
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
df.info()
df.describe()
```

```python
df.groupby("species").aggregate(
    avg_sep_len = ("sepal_length","mean"),
    max_petal_width = ("petal_width","max"),
    std_sep_width = ("sepal_width","std")
)
```

```python
df2 = df.copy()
df2 = df2.assign(petal_ratio = df["petal_length"]/df["petal_width"])
```

```python
df2.groupby("species").agg(
    avg_petal_ratio = ("petal_ratio","mean")
)
```
